내장함수

```python
##
# 이 프로그램은 사람들의 정보를 정렬하여 출력한다.
#
class Person(object):
    def __init__(self, name, age):
            self.name = name
            self.age = age
    def __repr__(self):
            return "<이름: %s, 나이: %d>" % (self.name, self.age)

def KeyAge(person):
    return person.age

people = [Person("홍길동", 25), Person("김철수", 30), Person("이영희", 22)]
print(sorted(people, key=KeyAge))                
```
