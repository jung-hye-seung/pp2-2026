```python
# 20261014 실습
# 이 프로그램은 사람들의 정보를 정렬하여 출력한다.
#
class Person(object):
    def __init__(self, name, age):
            self.name = name
            self.age = age
    def __repr__(self):
            return "<이름: %s, 나이: %d>" % (self.name, self.age)

def KeyAge(person):
    return person.age

people = [Person("오세훈", 25), Person("안철수", 30), Person("이영희", 22)]
print(sorted(people, key=KeyAge))                
```

```python
# 20261014 실습

def celsius(T):
    return (5/9) * (T - 32.0)
f_temp =[0, 10, 20, 30, 40, 50]

c_temp = map(celsius, f_temp)
print(list(c_temp))
```
