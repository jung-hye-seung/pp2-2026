타입힌트는 함수의 매개변수와 반환값의 타입을 저장하는 목적으로 사용하며, 엄격한 타입지정으로 인한 프로그래머의 의도를 명확히 전달하기 위함이다. 안써도 되지만 파이썬 13부터는 쓰는 것을 권장하고 있다.

```python
def greet(msg):
    print("in greet:", msg)
    return msg + "!"

print(greet("안녕"))
greet(3.141592)
```
