# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1

class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()
        Base1.display(self)
        Base2.displat2(self)

def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4,5,6)
    c2.print()

if __name__=="__main__":
    test()
```
