# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self):
        self.attr1 = 10

class Base2:
     def __init__(self):
         self.attr2 = 20

class C(Base1, Base2):
    def __init__(self):
        Base1.__init__(self)
        Base2.__init__(self)
        self.attr3 = 30

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")

def test():
    c = C()
    c.print()

if __name__=="__main__":
    test()
```
